# AstroWave: del dato astronómico al sonido

## Principio del proyecto

AstroWave se plantea como un proyecto de **sonificación de datos astronómicos**.

La sonificación consiste en utilizar sonido para representar información
contenida en datos. En este contexto, una variable astronómica no se convierte
directamente en sonido: necesitamos definir una regla que determine cómo será
representada mediante uno o más parámetros acústicos.

Por ejemplo:

$$
T \longrightarrow g(T) \longrightarrow f_{\mathrm{sound}}
$$

donde:

- $T$ es la temperatura de una estrella;
- $g(\cdot)$ es una transformación que nosotros diseñamos;
- $f_{\mathrm{sound}}$ es una frecuencia sonora utilizada para representar
  esa temperatura.

La temperatura **no es** una frecuencia acústica. La función $g$ constituye
una decisión de modelación.

---

## Sonificación versus creación sonora inspirada en datos

Es útil distinguir dos enfoques posibles.

### Sonificación de datos

En una sonificación, el sonido busca representar alguna propiedad o relación
presente en los datos.

Por ejemplo, si dos estrellas tienen temperaturas similares, podríamos
esperar que su representación sonora también conserve algún tipo de
proximidad perceptible.

El objetivo principal no es necesariamente que el resultado sea musicalmente
agradable, sino que exista una relación explícita y reproducible entre los
datos y el sonido.

### Creación sonora inspirada en datos

También es posible utilizar datos astronómicos como inspiración para una
composición, una textura sonora o una experiencia artística.

En ese caso, las decisiones pueden responder principalmente a criterios
estéticos y no es necesario que las relaciones entre los datos originales
se conserven en el sonido.

Ambos enfoques son válidos, pero responden a preguntas diferentes.

**AstroWave utilizará la sonificación como principio rector**, aunque las
decisiones de diseño sonoro también podrán considerar aspectos perceptuales
y estéticos.

---

## ¿Qué significa modelar en AstroWave?

Modelar no significa necesariamente construir una ecuación complicada ni
utilizar aprendizaje de máquina.

En este proyecto, modelar comienza simplemente al definir una transformación:

$$
y = g(x)
$$

donde una variable astronómica $x$ se transforma en un parámetro acústico $y$.

Por ejemplo:

$$
f_{\mathrm{sound}} = g(T)
$$

$$
\tau_{\mathrm{delay}} = h(d)
$$

$$
A = q(L)
$$

donde $T$, $d$ y $L$ podrían representar temperatura, distancia y
luminosidad.

**Cada función constituye una hipótesis de representación.**

Por eso, antes de elegir una transformación deberíamos preguntarnos:

- ¿qué propiedad de los datos queremos conservar?
- ¿qué rango tienen los valores originales?
- ¿la relación debe conservar el orden?
- ¿valores astronómicos próximos deberían producir sonidos próximos?
- ¿el parámetro acústico resultante se encuentra en un rango útil o perceptible?
- ¿qué ocurre con valores extremos?

Estas preguntas forman parte del proceso de modelación.

---

## Pipeline conceptual

El proyecto puede entenderse inicialmente como:

$$
\text{datos astronómicos}
\longrightarrow
\text{exploración}
\longrightarrow
\text{modelo de mapeo}
\longrightarrow
\text{parámetros acústicos}
\longrightarrow
\text{sonido}
$$

Solo después de comprender este proceso será necesario decidir cómo presentar
la sonificación mediante una interfaz.

---

## Principios de trabajo

Durante el desarrollo de AstroWave intentaremos mantener cuatro principios:

1. **Datos reales**  
   Las variables utilizadas deben provenir de fuentes astronómicas
   identificables y reproducibles.

2. **Mapeos explícitos**  
   Toda transformación entre una variable astronómica y un parámetro acústico
   debe poder escribirse y explicarse.

3. **Relaciones interpretables**  
   Cuando sea posible, la sonificación debería conservar alguna relación
   relevante presente en los datos originales.

4. **Reproducibilidad**  
   Dados los mismos datos y las mismas reglas de mapeo, el procedimiento
   debe producir el mismo resultado.

Estos principios no determinan cómo debe sonar AstroWave. Su propósito es
ayudarnos a justificar las decisiones que iremos tomando.

---

**El sonido no es el dato: es una representación del dato.**

## 2. Solicitar explícitamente nuevas propiedades

La consulta básica nos permitió identificar correctamente a Betelgeuse, pero no
entregó automáticamente todas las variables que podrían interesarnos para una
sonificación.

Ahora ampliaremos la consulta solicitando explícitamente algunas propiedades:

- **tipo de objeto**;
- **tipo espectral**;
- **paralaje**.

Este paso ilustra una idea importante en ciencia de datos:

$$
\text{pregunta del proyecto}
\longrightarrow
\text{variables necesarias}
\longrightarrow
\text{campos disponibles en la fuente}
$$

No debemos asumir que una base de datos contiene exactamente las variables que
imaginamos al formular el proyecto.

````python
from astroquery.simbad import Simbad

simbad = Simbad()

simbad.add_votable_fields(
    "otype",
    "sp_type",
    "parallax",
)

betelgeuse_extended = simbad.query_object("Betelgeuse")

betelgeuse_extended
````

### Cómo leer esta tabla sin perdernos

La consulta extendida contiene muchas columnas, pero **no todas cumplen el mismo
papel**. Para comenzar, podemos organizarlas en cuatro familias.

#### 1. Identidad: ¿qué objeto estamos mirando?

- `main_id`: identificador principal utilizado por SIMBAD.
- `matched_id`: nombre que coincidió con nuestra búsqueda.

En este caso, `* alf Ori` y `NAME Betelgeuse` se refieren al mismo objeto:
**Betelgeuse = Alpha Orionis**.

#### 2. Posición: ¿dónde está en el cielo?

- `ra`: ascensión recta.
- `dec`: declinación.

Juntas permiten ubicar el objeto sobre la esfera celeste. Los campos
`coo_err_maj`, `coo_err_min` y `coo_err_angle` describen la incertidumbre de
esa posición; `coo_wavelength` indica la región espectral de origen de las
coordenadas y `coo_bibcode` identifica su referencia bibliográfica.

No necesitamos sonificar estos metadatos, pero nos recuerdan algo importante:

> un dato científico no es solo un número; también tiene **incertidumbre,
> procedencia y contexto de medición**.

#### 3. Clasificación: ¿qué clase de objeto es?

- `otype`: tipo de objeto astronómico.
- `sp_type`: tipo espectral.

Para Betelgeuse aparecen:

```text
otype   = s*r
sp_type = M1-M2Ia-Iab
```

SIMBAD utiliza `s*r` como código compacto para una **supergigante roja**.

El tipo espectral `M1-M2Ia-Iab` puede leerse, de manera simplificada, así:

```text
M1-M2    |    Ia-Iab
   ↓             ↓
tipo y       clase de
subtipo      luminosidad
espectral    (supergigante)
```

- **M** indica una estrella de clase espectral M, asociada a temperaturas
  relativamente bajas y apariencia rojiza;
- **1–2** indica un subtipo dentro de la clase M;
- **Ia–Iab** corresponde a clases de luminosidad de estrellas supergigantes
  muy luminosas.

No necesitamos memorizar esta nomenclatura. Lo importante para AstroWave es
reconocer que `sp_type` **contiene información física** y, por tanto, podría
llegar a participar en una decisión de sonificación.

#### 4. Paralaje: una medición relacionada con la distancia

Los campos principales son:

- `plx_value`: valor medido de la paralaje;
- `plx_err`: incertidumbre asociada;
- `plx_qual`: indicador de calidad;
- `plx_bibcode`: referencia bibliográfica de la medición.

Para Betelgeuse:

$$
\pi = 6.55\ \mathrm{mas},
\qquad
\sigma_\pi = 0.83\ \mathrm{mas}
$$

donde `mas` significa **milisegundos de arco**.

La paralaje es un pequeño desplazamiento angular aparente que puede utilizarse
para obtener información sobre la distancia a una estrella. Por ahora **no la
convertiremos todavía en distancia**: primero queremos comprender qué mide y
qué incertidumbre posee.

Los campos `plx_prec` y `plx_err_prec` son metadatos técnicos sobre la
precisión con que se representan esos valores en la tabla; no son nuevas
propiedades físicas de Betelgeuse.

---

#### ¿Qué nos interesa para AstroWave?

Esta tabla ya permite distinguir entre:

**Propiedades o mediciones que podrían contribuir a una representación sonora**
```text

(y más adelante: temperatura, luminosidad...)
```

y **metadatos necesarios para comprender y auditar los datos**
```text

referencia bibliográfica

```

**No todas las columnas tienen que convertirse en sonido.** Elegir qué
variables representar, y cómo hacerlo, forma parte de la modelación del

## 3. ¿Qué variables necesita realmente AstroWave?

Hasta ahora aprendimos a consultar e interpretar algunas propiedades de
Betelgeuse. El siguiente paso no es generar sonido, sino decidir **qué
variables físicas podrían ser relevantes para el proyecto** y comprobar si

Para evitar mezclar preguntas distintas, separaremos tres niveles:

$$
\text{significado físico}
\longrightarrow
\text{disponibilidad del dato}

La última columna contiene solamente **hipótesis de trabajo**. No hemos

| Variable candidata | Significado físico | Unidad / tipo | Estado actual | Posible papel acústico |
|---|---|---|---|---|
| Temperatura efectiva | Estado térmico de la estrella | K | Aún no consultada | Pitch o brillo espectral |
| Distancia | Separación respecto del observador | pc, ly | Debemos derivarla u obtenerla de otra fuente | Espacialidad, delay u otro parámetro |

### Una distinción importante

Que una variable esté disponible **no significa que debamos sonificarla**.

Por ejemplo, la paralaje puede ser muy valiosa para estimar una distancia,
pero quizá sea la **distancia** —y no la paralaje misma— la propiedad que
queramos representar.

De manera similar:

$$
\text{dato disponible}
\neq
\text{variable que queremos comunicar}
\neq
\text{parámetro acústico}
$$

Esta separación nos ayudará a evitar mappings arbitrarios.

````python
campos_simbad = Simbad.list_votable_fields()

campos_simbad
````

### Buscar sin recorrer manualmente 116 opciones

SIMBAD ofrece muchas posibilidades de consulta. En lugar de revisar una por
una, podemos buscar palabras relacionadas con las variables que necesita
AstroWave.

Esta búsqueda no recupera todavía valores de Betelgeuse. Estamos inspeccionando
el **esquema de la fuente de datos**:

$$
\text{variable que necesitamos}
\longrightarrow
\text{¿existe un campo relacionado en SIMBAD?}
$$

Buscaremos inicialmente términos asociados a:

- temperatura;
- luminosidad;
- distancia;
- paralaje.

````python
campos_df = campos_simbad.to_pandas()

terminos = r"temp|temperature|lum|lumin|dist|parallax"

mask = (
    campos_df["name"]
    .astype(str)
    .str.contains(terminos, case=False, regex=True, na=False)
    |
    campos_df["description"]
    .astype(str)
    .str.contains(terminos, case=False, regex=True, na=False)
)

campos_relevantes = (
    campos_df.loc[mask, ["name", "description", "type"]]
    .reset_index(drop=True)
)

campos_relevantes
````

### Un hallazgo importante: SIMBAD también almacena distancias

Nuestra búsqueda encontró dos tipos de información relacionados con la
distancia:

- `parallax`: conjunto de campos básicos de paralaje;
- `mesPLX`: colección de mediciones de paralaje;
- `mesDistance`: colección de estimaciones de distancia publicadas.

Esto introduce otra distinción importante:

$$
\text{valor resumido o seleccionado}
\neq
\text{colección de mediciones publicadas}
$$

Una misma estrella puede tener varias estimaciones de distancia obtenidas con
métodos o publicaciones diferentes.

Para AstroWave no deberíamos elegir una de ellas automáticamente: primero
debemos conocer qué contiene la colección y qué procedencia tiene cada valor.

````python
distance_columns = Simbad.list_columns("mesDistance")

distance_columns
````

### ¿Qué contiene una medición de distancia?

La colección `mesDistance` no almacena solamente un número. Para cada
estimación puede conservar información como:

- `dist`: valor de distancia;
- `unit`: unidad de la medida (`pc`, `kpc` o `Mpc`);
- `minus_err` y `plus_err`: incertidumbres inferior y superior;
- `method`: método utilizado para estimar la distancia;
- `qual`: indicador de calidad;
- `bibcode`: referencia bibliográfica de la medición.

Esto es importante para AstroWave porque una misma estrella puede tener
**más de una estimación de distancia**.

Por tanto, antes de utilizar una distancia en un modelo de sonificación
debemos conocer su procedencia y decidir qué criterio utilizaremos para
seleccionarla.

````python
simbad_distance = Simbad()

simbad_distance.add_votable_fields("mesDistance")

betelgeuse_distances = simbad_distance.query_object("Betelgeuse")

betelgeuse_distances
````

### Un resultado también puede ser la ausencia de un dato

La colección `mesDistance` existe en SIMBAD, pero la consulta realizada para
Betelgeuse no devuelve una medición de distancia en esos campos.

Los símbolos `--` indican valores ausentes.

Esto nos permite distinguir tres situaciones:

$$
\text{la variable existe en la fuente}
\neq
\text{la variable está disponible para este objeto}
\neq
\text{tenemos un valor utilizable para el proyecto}
$$

Por tanto, la ausencia de un valor **no es un error del notebook**. Es un
resultado de nuestra auditoría.

Para AstroWave esto significa que, si queremos utilizar distancia, tendremos
que evaluar otra forma reproducible de obtenerla: por ejemplo, derivarla a
partir de una medición apropiada o consultar otra fuente astronómica.

````python
distance_fields = [
    "mesdistance.dist",
    "mesdistance.unit",
    "mesdistance.minus_err",
    "mesdistance.plus_err",
    "mesdistance.method",
    "mesdistance.qual",
    "mesdistance.bibcode",
]

betelgeuse_distances[distance_fields]
````

````python
for field in distance_fields:
    print(
        f"{field:28s} "
        f"masked = {bool(betelgeuse_distances[field].mask[0])}"
    )
````

### Conclusión de esta auditoría

La consulta confirma que `mesDistance` existe como colección en SIMBAD, pero

Algunos campos asociados aparecen sin un valor científicamente utilizable.
Por eso, para nuestro propósito, la pregunta fundamental no es solamente si

En esta consulta, para la distancia almacenada en `mesDistance`, la respuesta

Esto no significa que SIMBAD sea una fuente inadecuada. Al contrario, ya nos

### 5.1. Buscar a Betelgeuse en Gaia a partir de su posición

Ya conocemos la posición de Betelgeuse a partir de SIMBAD.

En lugar de buscar manualmente un identificador de Gaia, utilizaremos esas
coordenadas para realizar una búsqueda espacial alrededor de la estrella.

El procedimiento será:

$$
\text{coordenadas SIMBAD}
\longrightarrow
\text{búsqueda espacial en Gaia DR3}
\longrightarrow
\text{fuentes candidatas}
\longrightarrow
\text{auditoría de la identificación}
$$

Es importante no asumir que la fuente más cercana es automáticamente
Betelgeuse. Primero inspeccionaremos los candidatos y su separación angular.

En este paso **no utilizaremos todavía temperatura, distancia ni luminosidad**.
Primero necesitamos identificar correctamente el objeto.

````python
import astropy.units as u
from astropy.coordinates import SkyCoord
from astroquery.gaia import Gaia

# Coordenadas que ya obtuvimos desde SIMBAD
ra_betelgeuse = float(betelgeuse_extended["ra"][0])
dec_betelgeuse = float(betelgeuse_extended["dec"][0])

betelgeuse_coord = SkyCoord(
    ra=ra_betelgeuse * u.deg,
    dec=dec_betelgeuse * u.deg,
    frame="icrs",
)

print("Coordenadas utilizadas para la búsqueda:")
print(f"RA  = {ra_betelgeuse:.8f} deg")
print(f"Dec = {dec_betelgeuse:.8f} deg")
````

````python
# Buscaremos fuentes de Gaia DR3 dentro de 60 segundos de arco
# alrededor de la posición proporcionada por SIMBAD.

search_radius = 60 * u.arcsec

Gaia.ROW_LIMIT = 100

job = Gaia.cone_search_async(
    betelgeuse_coord,
    radius=search_radius,
    table_name="gaiadr3.gaia_source",
)

gaia_candidates = job.get_results()

print(f"Número de fuentes candidatas: {len(gaia_candidates)}")
````

````python
print("Unidad de RA :", gaia_candidates["ra"].unit)
print("Unidad de Dec:", gaia_candidates["dec"].unit)
````

````python
# Gaia ya devuelve ra y dec con unidades angulares.
# No debemos multiplicarlas nuevamente por u.deg.

candidate_coords = SkyCoord(
    ra=gaia_candidates["ra"],
    dec=gaia_candidates["dec"],
    frame="icrs",
)

# Separación angular entre la posición SIMBAD de Betelgeuse
# y cada fuente encontrada en Gaia.
separations = betelgeuse_coord.separation(candidate_coords)

gaia_candidates["separation_arcsec"] = separations.arcsec

# Ordenar desde la fuente angularmente más próxima.
gaia_candidates.sort("separation_arcsec")

candidate_columns = [
    "source_id",
    "ra",
    "dec",
    "separation_arcsec",
    "phot_g_mean_mag",
    "parallax",
    "parallax_error",
    "pmra",
    "pmdec",
    "ruwe",
]

gaia_candidates[candidate_columns][:10]
````

### 5.2. TESS Input Catalog: una asociación explícita

SIMBAD identifica a Betelgeuse como `TIC 269273552`.

El **TESS Input Catalog (TIC)** contiene parámetros estelares y permite
consultas mediante identificadores explícitos. Entre sus campos existen
variables potencialmente relevantes para AstroWave, como:

- `Teff`: temperatura efectiva, en K;
- `d`: distancia, en pc;
- `lum`: luminosidad, en unidades solares;
- `plx`: paralaje, en mas;

junto con campos de incertidumbre asociados, como `e_Teff`, `e_d`, `e_lum` y
`e_plx`.

Que esos campos existan en el TIC **no garantiza todavía que Betelgeuse tenga
valores válidos en todos ellos**. Aplicaremos la misma regla que antes:

$$
\text{identificador explícito}
\longrightarrow
\text{consultar el registro}
\longrightarrow
\text{inspeccionar valores y faltantes}
\longrightarrow
\text{auditar su procedencia}
$$

Primero recuperaremos programáticamente el identificador TIC desde la lista de
SIMBAD y luego consultaremos el catálogo. Todavía no seleccionaremos ninguna
variable para la sonificación.



Si el registro contiene una temperatura efectiva utilizable, la emplearemos
más adelante **solo para una demostración sonora mínima**. Esa demostración no
definirá la solución de AstroWave; servirá únicamente para recorrer una vez el
pipeline completo desde un dato real hasta un sonido.

## 6. Una primera sonificación: del dato al sonido

Cerraremos este cuadernillo con **un solo ejemplo audible**.

El objetivo no es definir cómo debe sonar AstroWave, sino recorrer una vez el
pipeline completo con una sola variable:

$$
T
\longrightarrow
g(T)
\longrightarrow
f_{\mathrm{sound}}
\longrightarrow
\text{audio}
$$

Si el registro TIC contiene una temperatura efectiva (`Teff`) utilizable,
la transformaremos en una frecuencia audible.

Para esta demostración elegiremos deliberadamente:

$$
T_{\min}=3000\ \mathrm{K},
\qquad
T_{\max}=10000\ \mathrm{K}
$$

y un intervalo sonoro de dos octavas:

$$
f_{\min}=220\ \mathrm{Hz},
\qquad
f_{\max}=880\ \mathrm{Hz}.
$$

Primero normalizamos la temperatura:

$$
x=
\frac{T-T_{\min}}
     {T_{\max}-T_{\min}},
$$

y luego utilizamos una transformación exponencial:

$$
f_{\mathrm{sound}}
=
f_{\min}
\left(
\frac{f_{\max}}{f_{\min}}
\right)^x.
$$

Los límites de temperatura y frecuencia son **decisiones didácticas para este
ejemplo**. No son propiedades universales de las estrellas ni constituyen el
mapeo definitivo de AstroWave.

Esta función es, nuevamente, una **hipótesis de representación**.


````python
import numpy as np
from IPython.display import Audio, display

# Solo generaremos audio si el TIC contiene una Teff utilizable.
if "Teff" not in tic_betelgeuse.colnames:
    print("El registro TIC no contiene una columna Teff.")
    print("No generaremos audio a partir de una temperatura inexistente.")

else:
    teff_raw = tic_betelgeuse["Teff"][0]

    if np.ma.is_masked(teff_raw):
        print("La temperatura efectiva está ausente en este registro TIC.")
        print("No generaremos audio a partir de un valor faltante.")

    else:
        T = float(teff_raw)

        if not np.isfinite(T):
            print("La temperatura efectiva no contiene un valor finito.")
            print("No generaremos audio a partir de un valor no utilizable.")

        else:
            # Intervalo didáctico de temperatura para esta demostración.
            T_min = 3000.0
            T_max = 10000.0

            # Dos octavas audibles.
            f_min = 220.0
            f_max = 880.0

            # Normalización y limitación al intervalo [0, 1].
            x = np.clip(
                (T - T_min) / (T_max - T_min),
                0.0,
                1.0,
            )

            # Mapping temperatura -> frecuencia.
            f_sound = f_min * (f_max / f_min) ** x

            print(f"Temperatura efectiva:       {T:.1f} K")
            print(f"Valor normalizado:          {x:.4f}")
            print(f"Frecuencia representativa: {f_sound:.1f} Hz")

            # Síntesis de un tono sinusoidal breve.
            sample_rate = 44100
            duration_s = 2.5

            t = np.arange(
                int(sample_rate * duration_s)
            ) / sample_rate

            signal = 0.15 * np.sin(
                2 * np.pi * f_sound * t
            )

            # Fade-in / fade-out de 30 ms para evitar clics.
            fade_samples = int(0.03 * sample_rate)

            envelope = np.ones_like(signal)
            envelope[:fade_samples] = np.linspace(
                0.0, 1.0, fade_samples
            )
            envelope[-fade_samples:] = np.linspace(
                1.0, 0.0, fade_samples
            )

            signal *= envelope

            display(
                Audio(
                    signal,
                    rate=sample_rate,
                    normalize=False,
                )
            )
````

### 6.1. De una a dos variables: paralaje y reverberación

Hasta ahora utilizamos una sola relación:

$$
T_{\mathrm{eff}}
\longrightarrow
f_{\mathrm{sound}}.
$$

Ahora añadiremos una segunda dimensión perceptual utilizando la paralaje
recuperada desde SIMBAD.

La paralaje es una medida angular relacionada inversamente con la distancia:
en términos generales, **una paralaje menor corresponde a un objeto más
lejano**.

Para esta demostración construiremos:

$$
\pi
\longrightarrow
h(\pi)
\longrightarrow
\text{cantidad de reverberación}.
$$

Por tanto, tendremos dos mappings independientes:

$$
\boxed{
\begin{aligned}
T_{\mathrm{eff}} &\longrightarrow \text{frecuencia},\\
\pi &\longrightarrow \text{reverberación}.
\end{aligned}
}
$$

Usaremos un intervalo didáctico de paralaje entre 1 y 100 mas y una escala
logarítmica. Dentro de ese intervalo, una paralaje menor producirá una mayor
proporción de señal reverberada.

> **Importante:** la reverberación no es una propiedad física de Betelgeuse ni
> representa propagación real de sonido por el espacio. Es una **metáfora
> perceptual** que elegimos para demostrar cómo una segunda variable puede
> controlar otro atributo acústico.

Esta elección es, nuevamente, una **hipótesis de representación**.


````python
# Segunda variable:
# paralaje SIMBAD -> proporción de reverberación.

wet_mix = None

if "plx_value" not in betelgeuse_extended.colnames:
    print("La consulta SIMBAD no contiene `plx_value`.")

else:
    plx_raw = betelgeuse_extended["plx_value"][0]

    if np.ma.is_masked(plx_raw):
        print("La paralaje está ausente en la consulta SIMBAD.")

    else:
        plx_mas = float(plx_raw)

        if not np.isfinite(plx_mas) or plx_mas <= 0:
            print("La paralaje no contiene un valor utilizable.")

        else:
            # Incertidumbre, solo para informar si está disponible.
            plx_err_mas = None
            if "plx_err" in betelgeuse_extended.colnames:
                plx_err_raw = betelgeuse_extended["plx_err"][0]
                if (
                    not np.ma.is_masked(plx_err_raw)
                    and np.isfinite(float(plx_err_raw))
                ):
                    plx_err_mas = float(plx_err_raw)

            # Intervalo didáctico de referencia para la paralaje.
            plx_min = 1.0
            plx_max = 100.0

            # Escala logarítmica:
            # menor paralaje -> mayor proxy perceptual de lejanía.
            plx_clipped = np.clip(
                plx_mas,
                plx_min,
                plx_max,
            )

            distance_proxy = 1.0 - (
                np.log10(plx_clipped) - np.log10(plx_min)
            ) / (
                np.log10(plx_max) - np.log10(plx_min)
            )

            distance_proxy = np.clip(
                distance_proxy,
                0.0,
                1.0,
            )

            # Mantener la reverberación moderada para no ocultar el pitch.
            wet_mix = 0.1 + 0.70 * distance_proxy # 0.10 + 0.40 * distance_proxy

            if plx_err_mas is None:
                print(f"Paralaje SIMBAD:             {plx_mas:.2f} mas")
            else:
                print(
                    "Paralaje SIMBAD:             "
                    f"{plx_mas:.2f} ± {plx_err_mas:.2f} mas"
                )

            print(
                f"Proxy perceptual de lejanía: {distance_proxy:.4f}"
            )
            print(
                f"Mezcla reverberada (wet):    {100 * wet_mix:.1f} %"
            )
````

````python
# Comparación A/B:
# A) temperatura -> pitch
# B) temperatura -> pitch + paralaje -> reverberación
#
# La "reverberación" se implementa aquí de forma transparente mediante
# varias reflexiones retardadas y decrecientes. No modela un recinto físico.

if wet_mix is None:
    print("No se generará la comparación porque no hay una paralaje utilizable.")

elif "signal" not in globals():
    print(
        "Primero ejecuta la celda de la sonificación basada en temperatura."
    )

else:
    dry_signal = signal.copy()

    # Reflexiones sintéticas: (retardo en segundos, ganancia relativa)
    #reflections = [
    #    (0.035, 0.55),
    #    (0.070, 0.38),
    #    (0.120, 0.28),
    #    (0.190, 0.20),
    #    (0.290, 0.14),
    #    (0.420, 0.10),
    #]
    reflections = [
        (0.050, 0.60),
        (0.120, 0.45),
        (0.220, 0.35),
        (0.380, 0.25),
        (0.600, 0.18),
        (0.900, 0.12),
    ]

    max_delay_samples = max(
        int(delay_s * sample_rate)
        for delay_s, _ in reflections
    )

    output_length = len(dry_signal) + max_delay_samples

    # Señal seca, extendida para poder comparar con la cola reverberante.
    dry_padded = np.zeros(output_length)
    dry_padded[:len(dry_signal)] = dry_signal

    # Componente reverberada.
    wet_component = np.zeros(output_length)

    for delay_s, gain in reflections:
        delay_samples = int(delay_s * sample_rate)

        wet_component[
            delay_samples:delay_samples + len(dry_signal)
        ] += gain * dry_signal

    # Igualar aproximadamente el pico del componente reverberado
    # al de la señal seca antes de realizar la mezcla.
    dry_peak = np.max(np.abs(dry_signal))
    wet_peak = np.max(np.abs(wet_component))

    if wet_peak > 0:
        wet_component *= dry_peak / wet_peak

    sonified_signal = (
        (1.0 - wet_mix) * dry_padded
        + wet_mix * wet_component
    )

    # Igualar el pico final al de la señal seca para que la comparación
    # se concentre en el efecto y no en una diferencia global de nivel.
    final_peak = np.max(np.abs(sonified_signal))

    if final_peak > 0:
        sonified_signal *= dry_peak / final_peak

    print("A. Temperatura solamente: pitch")
    display(
        Audio(
            dry_padded,
            rate=sample_rate,
            normalize=False,
        )
    )

    print("B. Temperatura + paralaje: pitch + reverberación")
    display(
        Audio(
            sonified_signal,
            rate=sample_rate,
            normalize=False,
        )
    )
````
